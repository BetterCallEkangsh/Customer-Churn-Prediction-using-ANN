```python
import pandas as pd
import pickle
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('datasets/Churn_Modelling.csv')
```

```python
df.head()
```

```python
# Removing unnecessary columns
df.drop(['RowNumber', 'CustomerId', 'Surname'], axis=1, inplace=True)
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
le = LabelEncoder()
ohe = OneHotEncoder()
```

```python
X = df.drop('Exited', axis=1)
y = df['Exited']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
X_train['Gender'] = le.fit_transform(X_train['Gender'])
```

```python
X_test['Gender'] = le.transform(X_test['Gender'])
```

```python
ohot_tr = ohe.fit_transform(X_train[['Geography']]).toarray()
ohot_te = ohe.transform(X_test[['Geography']]).toarray()
```

```python
X_train.drop('Geography', axis=1, inplace=True)
X_test.drop('Geography', axis=1, inplace=True)
```

```python
geo_encode_tr = pd.DataFrame(ohot_tr, columns=ohe.get_feature_names_out(), index=X_train.index)
geo_encode_te = pd.DataFrame(ohot_te, columns=ohe.get_feature_names_out(), index=X_test.index)

X_train = pd.concat([X_train, geo_encode_tr], axis=1)
X_test = pd.concat([X_test, geo_encode_te], axis=1)
```

```python
from sklearn.preprocessing import StandardScaler
ss = StandardScaler()
```

```python
X_train = ss.fit_transform(X_train)
X_test = ss.transform(X_test)
```

```python
with open('One_hot_encoder.pkl','wb') as file:
    pickle.dump(ohe,file)

with open('Standard_scaler.pkl','wb') as file:
    pickle.dump(ss,file)

with open('Label_Encoder.pkl','wb') as file:
    pickle.dump(le,file)
```

### ANN Implementation

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping, TensorBoard
import datetime
```

```python
X_train.shape[1],
```

```python
model = Sequential(
    [
        Dense(64, activation='relu', input_shape = (X_train.shape[1],)),   #I/P Layer + HL1
        Dense(32, activation='relu'), # HL2
        Dense(1, activation='sigmoid') #Output Layer
    ]
)
```

```python
model.summary()
```

```python
import tensorflow
opt = tensorflow.keras.optimizers.Adam(learning_rate = 0.01)
loss = tensorflow.keras.losses.BinaryCrossentropy()
```

```python
model.compile(optimizer=opt, loss=loss, metrics=['accuracy'])
```

```python
## Logging
log_dir = 'logs/fit/' + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorflow_callback = TensorBoard(log_dir=log_dir, histogram_freq=1)
```

```python
## Early stoppper

early_stopper_callback = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
```

```python
history = model.fit(
    X_train, y_train, validation_data = (X_test, y_test), epochs=100,
    callbacks = [tensorflow_callback, early_stopper_callback]
)
```

```python
model.save('model.h5')
```
